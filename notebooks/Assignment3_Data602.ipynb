```python
# Install necessary libraries (if not already installed)
!pip install imbalanced-learn

# Import required libraries
import numpy as np
from sklearn.datasets import make_classification
from imblearn.over_sampling import RandomOverSampler
import pandas as pd
from collections import Counter
```

```python
# Step 2: Create an imbalanced dataset
X_data, y_labels = make_classification(
    n_samples=10000,             # Total samples
    n_features=3,                # Total features
    n_informative=3,             # All features are informative
    n_redundant=0,               # No redundant features
    n_classes=4,                 # Total number of classes
    n_clusters_per_class=1,      # Single cluster per class
    weights=[0.7, 0.2, 0.08, 0.02],  # Class distribution: 70%, 20%, 8%, 2%
    random_state=42              # For reproducibility
)

# Step 3: Count and print the number of samples per class in the base dataset
class_counts_before = Counter(y_labels)
print("Class distribution before resampling:", class_counts_before)
```

```python
# Step 4: Resample the dataset using RandomOverSampler
oversampler = RandomOverSampler(random_state=42)
X_resampled, y_resampled = oversampler.fit_resample(X_data, y_labels)

# Step 5: Count and print the number of samples per class in the resampled dataset
class_counts_after = Counter(y_resampled)
print("Class distribution after resampling:", class_counts_after)
```
